```python
import pandas as pd

# 1. Membaca dataset awal
df = pd.read_csv("automobile_dataset.csv")

# 2. Mengecek kondisi awal
print("Jumlah data awal :", df.shape)
print("Total missing value :", df.isnull().sum().sum())
print("Jumlah duplicate :", df.duplicated().sum())


# 3. Menghapus baris yang memiliki missing value pada kolom price
# price digunakan sebagai target, sehingga baris tanpa price dihapus
df = df.dropna(subset=["price"])


# 4. Mengisi missing value pada kolom numerik
# menggunakan nilai rata-rata (mean)
mean_columns = [
    "normalized-losses",
    "bore",
    "stroke",
    "horsepower",
    "peak-rpm"
]

for col in mean_columns:
    df[col] = df[col].fillna(df[col].mean())


# 5. Mengisi missing value pada num-of-doors
# menggunakan nilai yang paling sering muncul (modus)
df["num-of-doors"] = df["num-of-doors"].fillna(
    df["num-of-doors"].mode()[0]
)


# 6. Menghapus data duplicate
df = df.drop_duplicates()


# 7. Mengecek hasil cleaning
print("\n=== HASIL CLEANING ===")
print("Jumlah data :", df.shape)
print("Total missing value :", df.isnull().sum().sum())
print("Jumlah duplicate :", df.duplicated().sum())


# 8. Mengecek missing value per kolom
print("\nMissing value per kolom:")
print(df.isnull().sum())


# 9. Menyimpan hasil sebagai Clean Dataset v1
df.to_csv("Clean Dataset v1.csv", index=False)

print("\nClean Dataset v1 berhasil dibuat!")
```
