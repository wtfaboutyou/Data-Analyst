```python
import pandas as pd

# 1. Memuat Dataset
df = pd.read_csv('SampleSuperstore.csv')

# Menampilkan informasi dasar (tipe data dan missing values)
print("--- Info Dataset ---")
df.info()

# 2. Analisis Statistik Dasar (Numerik)
print("\n--- Ringkasan Statistik Dasar ---")
# Fungsi describe() otomatis menghitung Mean, Min, Max, Standar Deviasi, dan Kuartil (25%, 50%/Median, 75%)
display(df.describe())

# 3. Menghitung Modus
print("\n--- Modus Variabel Numerik ---")
display(df[['Sales', 'Quantity', 'Discount', 'Profit']].mode())

# 4. Praktik Exploratory Data Analysis (Kategorikal)
# Menggunakan value_counts() untuk melihat proporsi
print("\n--- Transaksi per Kategori (value_counts) ---")
display(df['Category'].value_counts())

# Menggunakan unique() untuk melihat nilai unik dalam bentuk array
print("\n--- Daftar Kategori Unik (unique) ---")
display(df['Category'].unique())

# Menggunakan nunique() untuk menghitung total nilai unik
print("\n--- Jumlah Sub-Kategori Unik (nunique) ---")
print(f"Total Sub-Kategori: {df['Sub-Category'].nunique()}")
```
